## Data Cleaning

### 7. Remove all rows where the `YIELD_KG` column has missing (`NaN`) values

```python
crop_df = crop_df.dropna(subset=['YIELD_KG']).copy()
print('Remaining rows after dropping missing YIELD_KG:', len(crop_df))
```

### 8. Fill missing values in the `AREA_ACRES` column with the mean of that column

```python
crop_df['AREA_ACRES'] = crop_df['AREA_ACRES'].fillna(crop_df['AREA_ACRES'].mean())
```

### 9. Replace missing values in the `DISTRICT` column with `'Unknown'`

```python
crop_df['DISTRICT'] = crop_df['DISTRICT'].fillna('Unknown')
```

### 10. Replace missing values in the `FERTILIZER_USED` column with `'No'`

```python
crop_df['FERTILIZER_USED'] = crop_df['FERTILIZER_USED'].fillna('No')
```

### 11. Replace missing values in the `YIELD_VALUE` column with the mean of the column

```python
crop_df['YIELD_VALUE'] = crop_df['YIELD_VALUE'].fillna(crop_df['YIELD_VALUE'].mean())
```

### 12. Convert the `HARVEST_DATE` column to datetime format (`%d/%m/%Y`)

```python
crop_df['HARVEST_DATE'] = pd.to_datetime(crop_df['HARVEST_DATE'], errors='coerce', format='%d/%m/%Y')
```

### 13. Remove the letter `'F'` from the `FARM_ID` column and convert to integer

```python
crop_df['FARM_ID'] = crop_df['FARM_ID'].str.replace('F', '', regex=False).astype(int)
```

### 14. Re-check missing values in the data again after cleaning

```python
crop_df.isnull().sum()
```

### 15. Re-check the data types of all columns again after cleaning

```python
crop_df.dtypes
```

## Data Analysis and Visualization

### 16. Calculate the total yield (`YIELD_KG`) of each type of crop

```python
total_yield_by_crop = crop_df.groupby('CROP_TYPE').agg({'YIELD_KG': 'sum'})
total_yield_by_crop
```

### 17. Plot a bar chart of Total Yield by Crop Type (KG)

```python
plt.figure(figsize=(8, 5))
total_yield_by_crop['YIELD_KG'].plot(kind='bar', color='#2b5c8f', edgecolor='black')
plt.title('Total Yield by Crop Type (KG)')
plt.xlabel('Crop Type')
plt.ylabel('Yield (KG)')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

### 18. Create a pivot table summarizing total `YIELD_KG` by `HARVEST_DATE`

```python
yield_pivot = crop_df.pivot_table(values='YIELD_KG', index='HARVEST_DATE', aggfunc='sum')
yield_pivot.head(10)
```

### 19. Plot a line chart showing the yield trend over time and save as `'crop_yield_trend.png'`

```python
plt.figure(figsize=(12, 5))
plt.plot(yield_pivot.index, yield_pivot['YIELD_KG'], color='#1f77b4', linewidth=1)
plt.title('Yield Trend Over Time (KG)')
plt.xlabel('Harvest Date')
plt.ylabel('Yield (KG)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig('crop_yield_trend.png', dpi=300)
plt.show()
```
